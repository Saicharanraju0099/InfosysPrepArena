### FOR LOOP
**BASIC SYNTAX**

**for variable in sequence:**

   **# Code block to execute repeatedly**
   
   **# This block must be indented**

```python
items = (1,2,3,4,5)
for item in items:
    print(f"Rounds :{item}")
```

```python
items = "Python"
for item in items:
    print(f"Chars :{item}")
```
