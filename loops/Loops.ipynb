### BREAK STATEMENT
**The break statement in Python is a loop control statement used to immediately terminate the execution of the current loop (either a for or while loop) and transfer program control to the next line of code following the loop.**

```python
names = ["john", 'Maria', ' ', 'Kumar']
for name in names:
    if name == ' ':
        print('Empty Name Found!')
        break
    print(f'Name = {name}')
```

### Continue Statement
**The continue statement in Python is a loop control statement that skips the rest of the code inside the current loop iteration and immediately jumps to the next iteration. It is used within for and while loops to alter the normal flow of execution based on specific conditions**

```python
names = ["john", 'Maria', ' ', 'Kumar']
for name in names:
    if name == ' ':
        print('Empty Name Found!')
        continue
    print(f'Name = {name}')
```

### Pass Statement 
**The pass statement in Python is a null operation; nothing happens when it executes. It is used as a placeholder where a statement is syntactically required by Python's grammar, but you do not want any code to run yet.**

```python
names = ["john", 'Maria', ' ', 'Kumar']
for name in names:
    if name == ' ':
        print('Empty Name Found!')
        pass
        name = name.replace(' ', 'Unknown')
    
    print(f'Name = {name}')
```

```python
## loop through a list of days and print only the weekdays, skipping weekends using continue statement
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekends = ['Saturday', 'Sunday']
for day in days:
    if day in weekends:
        break
    print(f"Day: {day}")
```

### for else
**The else keyword in a for loop specifies a block of code to be executed when the loop is finished**


```python
items = [1,4,3,5,7]
for i in items:
    if i % 2 == 0:
        print(f"Even Number Found: {i}. Stopping the search.")
        break
else:
    print("All numbers are odd")
```

```python
# Check Whether any file name appear more than once
file_names = [
    "report.csv",
    "data.xlsx",
    "summary.docx",
    "report.csv",
    "data.csv"
]

for file in file_names:
    count = 1
    for compare_file in file_names:
        if file == compare_file:
            count += 1
    if count > 1:
        print(f"Duplicate File Found: {file}")
        break
else:
    print("No Duplicate Files Found")
    
```

### Nested Loops
**A nested loop in Python is a loop placed inside the body of another loop. This structure allows you to perform repeated actions over multi-dimensional data or for every combination of items across multiple sequences**

```python
for x in range(3):
    for y in range(2):
        for z in range(2):
            print(f"(x= {x}, y = {y}, z = {z})")
```

```python
colors = ['Red', 'Green', 'Blue']
sizes  = ['S', 'M', 'L']
for color in colors:
    for size in sizes:
        print(f"{color} - Size {size}")
```

```python
years = [2026,2027]
months = ['Jan', 'Feb']
days   = range(1, 29)

for y in years:
    for m in months:
        for d in days:
            print(f"report_{y}_{m}_{d}.csv")
            
```

```python
# SELECT Count(*) FROM customers where id IS NULL;
tables = ['customers', 'orders', 'products', 'prices']
columns = ['id','date_created']
for table in tables:
    for column in columns:
        print(f"SELECT Count(*) FROM {table} where {column} IS NULL;")
```

### WHILE LOOP
**A while loop in Python repeatedly executes a block of code as long as a specified condition remains True. This loop type is ideal when the number of iterations is not known in advance, such as when processing user input or waiting for a specific event.**

```python
# Build a counter that counts from 1 to 5 using a while loop
count = 1      ## Initialization
while count <= 10:     ## Condition Check
    print(f"Count is: {count}")
    count += 2    ## update statement

    
```

```python
# Write a program that keeps asking 
# Do you Agree ?
# Until User types 'yes'
response = ''
while response.lower() != 'yes':
    response = input("Do You Agree? (yes/no): ")
print("Thank's for agree!")
```

```python
## 3 Attempts to Login
# yes within three attempts -> 'Glad You Logged In'
# otherwise -> 'Sorry, Try Again Later'
attempts = 0
while attempts < 3:
    response = input("Do You Agree? (yes/no):")
    if response.lower() == 'yes':
        print("Glad You Logged In")
        break
    attempts += 1
else:
     print("sorry, Try Again Later")
```
