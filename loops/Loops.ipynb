### FOR LOOP
**BASIC SYNTAX**

**for variable in sequence:**

   **# Code block to execute repeatedly**
   
   **# This block must be indented**

```python
items = (1,2,3,4,5)
for item in items:
    print(f"Rounds :{item}")
```

```python
## Sequence as String
items = "Python"
for item in items:
    print(f"Chars :{item}")
```

```python
## Sequence as Range
for item in range(1, 10, 2):
    print(f"Rounds : {item}")
```

```python
scores = [90, 25, 67, 45, 80]
total = 0
for score in scores:
    total += score
    print(f"Total so far:", total)
print(f"final Total:", total)
```

```python
# Now we need to clean up the Files Data
# First Removing Spaces in files
files = ["Report 2020.pdf", " Data 2021.pptx", " Summary 2019.DOCX"]
for file in files:
    file_manupulation = file.strip().lower().replace('.pptx', '.csv')
    print(f"File after removing spaces: '{file_manupulation}'")
```

```python
# Print 7 table using for loop
for numbers in range(1, 11):
    result = 7 * numbers
    print(f"7 x {numbers} = {result}")
```

```python
## Print a left aligned pyramid of stars with 6 stars using a for loop
No_of_rows = int(input("Enter number of rows: "))
for i in range(No_of_rows):
    for j in range(i + 1):
        print("*", end="")
    print()
```
