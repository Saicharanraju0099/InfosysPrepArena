### Problem 1 — Sum of N numbers
**Read N then N integers; print their sum.**

**Input: N; a1 a2 ... aN**

**Output: sum**

```python
n = int(input("Enter the number of integers: "))
numbers = list(map(int, input().split()))
total = sum(numbers)
print(total)
```


### Problem 2 — Average of N 
**Read N and N integers; print average (2 decimals).**

**Input: N; a1 ... aN**

**Output: average formatted to 2 decimal places**

```python
n = int(input("Enter the number of integers: "))
data = []

for i in range(n):
    val = int(input())
    data.append(val)

    total = sum(data)
    print("Total =", total)

    avg = total / len(data)
    print("Average =", avg)
```

### Problem 4 — Count vowels and consonants
**Given a word; print number of vowels and consonants.**

**Output: two ints**

```python
sentence = "jaurney become destination"
count = 0

for char in 'aeiou':
    count += sentence.count(char)

print("Vowels Count:", count)
print("Sentence Length:", len(sentence))
```


### Problem 5 — Number palindrome check
**Given integer n; print 'Yes' if palindrome else 'No'.**

```python
n = input("Enter a number: ")

if n == n[::-1]:
    print("Yes")
else:
    print("No")
```

### Problem 8 — Factorial
**Input: n; Output: n! (iterative).**

```python
n = int(input("Enter n: "))
fact = 1

for i in range(1, n + 1):
    fact *= i

print(fact)
```


### Problem 9 — Fibonacci
**Input: n; Output: nth Fibonacci number (0-based, iterative).**

```python
n = int(input("Enter n: "))

a, b = 0, 1

for _ in range(n):
    a, b = b, a + b

print(a)
```


### Problem 10 — Prime check
**Input: n; Output: 'Prime' or 'Not Prime'.**

```python
n = int(input("Enter n: "))

if n < 2:
    print("Not Prime")
else:
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```


### Problem 11 — Unique list elements
Given a list; return unique elements in original order

```python
lst = [1, 2, 3, 2, 4, 1, 5]
unique = []

for item in lst:
    if item not in unique:
        unique.append(item)

print("Unique elements:", unique)
```

**Given list and k; rotate right by k positions.**

```python
lst = [1, 2, 3, 4, 5]
k = int(input("Enter k: "))

k = k % len(lst)  
rotated = lst[-k:] + lst[:-k]

print("Rotated list:", rotated)
```


### Problem 20 — Word frequency
**Given a sentence; print word->count.**

```python
sentence = "hello world hello python"
words = sentence.split()
freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

for word, count in freq.items():
    print(f"{word} -> {count}")
```
