### Dictionary

```python
my_dict = {'a':10,
            'b':20,
            'c':30,
            'd':40

}
print(my_dict)      ## ordered
print(my_dict['b']) ## Not index (keys)
## keys are unique
## values allow duplicates
```

### python dict

```python
person = {"id":101, "age":30, "city": "Mumbai"}
print(person.get("id"))
print(person.get("name"))

# Checks
print("age" in person)
print("name" in person)

## View Objects
print(person.keys())          
print(person.values())
print(person.items())
```

```python
## looping 
person = {"id":101, "age":30, "city": "Mumbai"}
for key, value in person.items():
    print(key,value)

```

### add remove update

```python
user = {"id":101, "age":30, "city": "Mumbai"}

## Add, Remove, Update
user["name"] = "charan"

## updating
user["age"] = 23
user.update({"age":21,"city":"delhi"})

## Remove
user.pop("name")
## remove last item
user.popitem()
print(user)
```

### Dict Creation

```python
## Dict Creation
user = {"id":None,
        "name":None,
        "age":None,
        "city":None
}

user = dict.fromkeys(["id","name","age","city"], None)
print(user)
```
