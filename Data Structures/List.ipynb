### Slicing


```python
list = ['a', 'b', 'c', 'd', 'e']
print(list[0:3])
print(list[2:])  # Output: ['c', 'd', 'e']
print(list[:3])  # Output: ['a', 'b', 'c']
print(list[:])   # Output: ['a', 'b', 'c', 'd', 'e']
```

### matrix slicing

```python
## matrix slicing
matrix = [
         ['a', 'b', 'c'],       ## Row 0
         ['d', 'e', 'f'],       ## Row 1
         ['g', 'h', 'i']        ## Row 2
]
print(matrix[0:2])  # Output: [['a', 'b', 'c'], ['d', 'e', 'f']]
print(matrix[2][:2])   # Output: ['g', 'h']
```

### Deep copy

```python
import copy
matrix = [
          ['a','b'],    #row 0
          ['c','d']     #row 1
]
matrix_copy = copy.deepcopy(matrix)
matrix.pop()
matrix_copy[0].append('z')
print("original: ",matrix)
print("copy :",matrix_copy)
```

```python
import copy
original = [
    ['a','b'],
    ['c','d']
]
copy1 = original
print("Same object :", original is copy1)
## Shallow copy
copy3 = copy.deepcopy(original)
print("Same object:",original is copy3)
print("Shared List:", original[0] is copy3[0])


```

### map()

```python
letters = ['a','b','c']
print(list(map(str.upper,letters)))

numbers = ['1','2','3']
print(list(map(int,numbers)))

names = ['Maria', 'John ', ' Kumar']
print(list(map(str.strip,names)))

names = ['Maria', 'John ', ' Kumar']
for n in map(str.strip,names):
    print(n)
```

### Filter()

```python
letters = ['a','','b',None,'c',False]
print(list(filter(None,letters)))

items = ['sql','123','python','42']
print(list(filter(str.isalpha,items)))

for i in filter(str.isalpha,items):
    print(i)
```
