### set{}

```python
my_set = {10,30,20}
## add
my_set.add(90)
# update
my_set |= {100,48}
# remove
my_set.discard(100)
print(my_set)
```

```python
## set methods
a = {10,20,30,40}
b = {30,40,50,60}
# mathematical operations
print(a.union(b))      ## unique values merging
print(a | b)
print(a.intersection(b))
print(a & b)
print(a.difference(b))
print(a - b)
```

 ### Set relationship

```python
a = {30,40}
b = {30,40,50,60}
print(a.issubset(b))
print(b.issuperset(a))
print(a.isdisjoint(b))
```
