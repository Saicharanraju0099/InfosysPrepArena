### Create a string made of the middle three characters
### Write a program to create a new string made of the middle three characters of an input string.

### Input

| Case   | Variable | Value        |
|--------|----------|--------------|
| Case 1 | str1     | "Saicharan"  |
| Case 2 | str2     | "Computers"  |

### Output

| Case   | Printed Output |
|--------|----------------|
| Case 1 | cha            |
| Case 2 | put            |


```python
def middle_three_chars(str1):
    print("Original String: ", str1)
    print("Length Of String is :",len(str1))
    middle = int(len(str1) // 2)
    print("middle Index Value is :",middle)
    ''' Slicing the string start : stop this return middle index value before char and middle index value after char last char is not included '''
    result = str1[middle - 1: middle + 2]              
    print("Middle Three Characters are :",result)       
middle_three_chars("Saicharan")
middle_three_chars("Computers")
```

### Given two strings, s1 and s2. Write a program to create a new string s3 by appending s2 in the middle of s1.
### Input

| Case   | Variable | Value    |
|--------|----------|----------|
| Case 1 | s1       | "Ault"   |
| Case 2 | s2       | "Kelly"  |

### Output

| Case   | Printed Output |
|--------|----------------|
| Case 1 | AuKellylt      |


```python
s1 = "Ault"
s2 = "Kelly"
# getting middle index of string s1
middle = int(len(s1)/2)
''' concatenating string s2 in middle of string s1 and last half of string s1 and first half of string s1 stored in result variable '''
result = s1[:middle] + s2 + s1[middle:] 
print("New String is:", result)
```

### Create a new string made of the first, middle, and last characters of each input string
### Given two strings, s1 and s2, write a program to return a new string made of s1 and s2’s first, middle, and last characters.
### Input

| Case   | Variable | Value      |
|--------|----------|------------|
| Case 1 | s1       | "America"  |
| Case 2 | s2       | "Japan"    |

### Output

| Case   | Printed Output |
|--------|----------------|
| Case 1 | AJrpan         |

```python
s1 = "America"
s2 = "Japan"
# Getting the first character of string s1 and s2
first_char = s1[0] + s2[0]
# Getting the middle character of string s1 and s2
middle = s1[len(s1)//2] + s2[len(s2)//2]
# getting the last character from s1 and s2
last_char = s1[-1] + s2[-1]
# Concatenating first, middle and last character of string s1 and s2
result = first_char + middle + last_char
print("New String is: ", result)
```

### Arrange string characters such that lowercase letters should come first
### Given string contains a combination of the lower and upper case letters. Write a program to arrange the characters of a string so that all lowercase letters should come first.
### Input

| Case   | Variable | Value      |
|--------|----------|------------|
| Case 1 | str1     | "PyNaTive" |

### Output

| Case   | Printed Output |
|--------|----------------|
| Case 1 | yaivePNT       |

```python
str1 = "PyNaTive"
print("Original String: ", str1)
## creating empty list to store uppercase and lowercase characters
upper_case = []
lower_case = []
# iterating through string and checking for uppercase and lowercase characters
for char in str1:
    if char.isupper():
        upper_case.append(char)
    else:
        lower_case.append(char)
 # concatenating uppercase and lowercase characters and storing in result variable
result = ''.join(lower_case + upper_case)
print("New String is: ", result)
```

### Count all letters, digits, and special symbols from a given string

### Input

| Case   | Variable | Value               |
|--------|----------|---------------------|
| Case 1 | str1     | "P@#yn26at^&i5ve"   |

### Output

| Case   | Chars | Digits | Symbols |
|--------|-------|--------|---------|
| Case 1 | 8     | 3      | 4       |

```python
def find_digits_chars_symbols(str1):
    char_count = 0
    digit_count = 0
    symbol_count = 0
    for char in str1:
        if char.isalpha():
            ## checking for alphabetic characters and counting them. char_count = char_count + 1 and storing in char_count variable
            char_count += 1
        elif char.isdigit():
            ## checking for digits and counting them. digit_count = digit_count + 1 and storing in digit_count variable
            digit_count += 1
        else:
            ## checking for symbol characters and counting them. symbol_count = symbol_count + 1 and storing in symbol_count variable
            symbol_count += 1
    print("characters=",char_count, "Digits=",digit_count,"Symbol=",symbol_count)
find_digits_chars_symbols("P@#yn26at^&i5ve")
```

### Create a mixed String using the following rules
### Given two strings, s1 and s2. Write a program to create a new string s3 made of the first char of s1, then the last char of s2, Next, the second char of s1 and second last char of s2, and so on. Any leftover chars go at the end of the result.
### Input

| Case   | Variable | Value  |
|--------|----------|--------|
| Case 1 | s1       | "Abc"  |
| Case 2 | s2       | "Xyz"  |

### Output

| Case   | Printed Output |
|--------|----------------|
| Case 1 | AzbycX         |

```python
s1 = "Abc"
s2 = "Xyz"
# getting the string length
s1_length = len(s1)
s2_length = len(s2)
# getting the length of bigger string
length = s1_length
if s1_length > s2_length:
    length = s1_length
else:
    length = s2_length
    result = ""
# Reverse the s2
s2 = s2[::-1]
# iterating through the string and concatenating the characters of s1 and s2
for i in range(length):
    if i < s1_length:
        result += s1[i]
        if i < s2_length:
            result += s2[i]
print("New String is: ", result)
```
