### Create a string made of the middle three characters
### Write a program to create a new string made of the middle three characters of an input string.

### Input

| Case   | Variable | Value        |
|--------|----------|--------------|
| Case 1 | str1     | "Saicharan"  |
| Case 2 | str2     | "Computers"  |

### Output

| Case   | Printed Output |
|--------|----------------|
| Case 1 | cha            |
| Case 2 | put            |


```python
def middle_three_chars(str1):
    print("Original String: ", str1)
    print("Length Of String is :",len(str1))
    middle = int(len(str1) // 2)
    print("middle Index Value is :",middle)
    result = str1[middle - 1: middle + 2]                        # Slicing the string start : stop this return middle index value before char
    print("Middle Three Characters are :",result)               # and middle index value after char last char is not included
middle_three_chars("Saicharan")
middle_three_chars("Computers")
```

### Write a program to create a new string made of an input string’s first, middle, and last character

```python
str1 = 'James'
print("original string is: ",str1)
# get first character from the string
first_char = str1[0]
#get middle character and length of string
l = len(str1)
middle = int(l/2)
# get middle character and add it to first character of string
first_middle = first_char + str1[middle]
# get last character and add it to first and middle character of string
result = first_middle + str1[l - 1] 
# print string
print("New String is: ", result)

```

### Given two strings, s1 and s2. Write a program to create a new string s3 by appending s2 in the middle of s1.
### Input

| Case   | Variable | Value    |
|--------|----------|----------|
| Case 1 | s1       | "Ault"   |
| Case 2 | s2       | "Kelly"  |

### Output

| Case   | Printed Output |
|--------|----------------|
| Case 1 | AuKellylt      |


```python
s1 = "Ault"
s2 = "Kelly"
# getting middle index of string s1
middle = int(len(s1)/2)
# concatenating string s2 in middle of string s1 and last half of string s1 and first half of string s1 stored in result variable
result = s1[:middle] + s2 + s1[middle:] 
print("New String is:", result)
```
