### Math Operators

```python
print(10 + 5)  # Addition
print(10 - 5)  # Subtraction
print(10 * 5)  # Multiplication
print(10 / 5)  # Division
print(10 // 3) # Floor Division
print(10 % 3)  # Modulus
print(10 ** 2) # Exponentiation
```

### Short Hand Operator

```python
## Sum
x = 2
# x = x + 3
x += 3 # This is equivalent to x = x + 3
print(x)
```

```python
## Substraction
x = 10
x -= 2 # This is equivalent to x = x - 2
print(x)
```

```python
## Multiplication
x = 5
x *= 2 # This is equivalent to x = x * 2
print(x)
```

```python
# Division
x = 10
x /= 2 # This is equivalent to x = x / 2
print(x)
```

```python
## Floor Division
x = 10
x //= 3 # This is equivalent to x = x // 3
print(x)
```

```python
## Modulus
x = 10
x %= 3 # This is equivalent to x = x % 3
print(x)
```

```python
# Exponentiation
x = 10
x **= 2 # This is equivalent to x = x ** 2
print(x)
```

## Rounding

```python
## Abs function returns the absolute value of a number without considering its sign.
# Measure the distance from one point to another
print(abs(2 - 10))
```

```python
## Rounding Numbers
price = 24.443
print(round(price)) # Rounds to the nearest integer
print(round(price, 1)) # Rounds to 1 decimal place
```

```python
### ceil and floor functions
import math
print(math.ceil(2.3))  # Rounds up to the nearest integer
print(math.floor(2.7)) # Rounds down to the nearest integer
```

## Validation


```python
## is_integer() method checks if the number is an integer or not.
x = 7.0
print(x.is_integer()) # True, because 7.0 is an integer value
y = 7.5
print(y.is_integer()) # False, because 7.5 is not an integer value
```

```python
## is_instance() function checks if a variable is an instance of a specific data type.
x = 10
print(isinstance(x, int))   # True, because x is an integer
print(isinstance(x, float)) # False, because x is not a float
```

```python
import random
print(random.randint(1, 101)) # Generates a random integer between 1 and 101 (inclusive)
random_number = random.randint(1, 101)
if random_number % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```
