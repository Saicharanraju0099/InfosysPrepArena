### len()
### count()

```python
password = "2335423Qfd"
print(len(password))
if len(password) < 8:
    print("Password is too short.")
elif len(password) > 20:
    print("Password is too long.")
else:
    print("Password is of valid length.")
```

```python
text = '''python is a high-level programming language.
it is widely used for web development, data analysis, artificial intelligence, and scientific computing.
python is powerful and easy to learn programming language.
many developers prefer python for its simplicity and versatility.'''
print(text.count("python"))
```
