## Slicing And Indexing

```python
text = "hello"
# Extract the first Characters
print(text[0])
print(text[-5])
# Extract the last character
print(text[4])
print(text[-1])
# Extract e 
print(text[1])
print(text[-4])
```

```python
date = "2026-02-18"
print(date[0:4])     # Extract The Year
print(date[5:7])     # Extract The Month
print(date[8:10])    # Extract The Day
```

### Data Cleaning
### Remove Spaces

```python
## WhiteSpace Cleanup
text = " DataScience"
text_1 = "Data Science...."
print(text.lstrip())
print(text.rstrip())
print(text.strip())
print(text_1.strip("."))
```

```python
text = " QuantumComputing"
print(len(text))
print(len(text.strip()))
whitespace_count = len(text) - len(text.strip())
is_cleaned = whitespace_count == 0
print(f"Number of leading whitespace characters: {whitespace_count}")
print(f"cleaned: {is_cleaned}")
```
