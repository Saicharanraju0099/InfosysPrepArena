### Class and Object(Fundatation)

```python
class Car:
    brand = "Toyota"     # Attribute data
    wheels = 4           # Attribute data

# Create an object of the class
car1 = Car()
car2 = Car()
car2.brand = "Honda"
car2.wheels = 4


# Access attributes
print(car1.brand)
print(car1.wheels)
print(car2.brand)
print(car2.wheels)
```
