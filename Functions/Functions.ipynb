### Lambda Function

```python
multiple = lambda x: x * 2
print(multiple(5))
```

```python
add = lambda x,y: x + y
print(add(1,2))
```

```python
check = lambda i: i in "python"
print(check("n"))
```

```python
prices = ['₹200.68', '₹150.50', '₹50.00', '₹900.35']

print(list(map(lambda p:float(p.replace('₹', '')),prices)))

```

### Filter  + Lambda

```python
prices = [120, 30, 450, 78, 90]
print(list(filter(lambda p: p >= 100, prices)))
```

```python
students = [["maria", 89],
            ["rohan", 56],
            ["susan", 78],
            ["daniel", 90]]
print(list(filter(lambda row:row [1] > 70, students)))

```

```python
students = [["maria", 89],
            ["rohan", 56],
            ["susan", 78],
            ["max", 90]]
print(list(filter(lambda row:row[0].startswith('m'), students)))
```
