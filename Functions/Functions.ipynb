### Global variable vs Local Variable

```python
case_rule = "lower"                  ## global variable
def clean_name(name):                ## Parameter
    cleaned = name.strip().lower()   ## local variable
    if case_rule == "lower":
        cleaned = cleaned.lower()
    print("Cleaned",cleaned)

clean_name(" MariA ")
print("The rule is:", case_rule)
```

### multiple inputs

```python
# Building Full Clean Name
def clean_name(first_name, last_name,country = "n/a"):
    first = first_name.strip().lower()
    last = last_name.strip().lower()
    full_name = first + " " + last
    print(full_name, "from", country)
clean_name("Sai "," Charan","India")      ## positional Arguments (un ordered)

clean_name(country="India",first_name="Sai",last_name="Charan")    ## Keyword Arguments (ordered)

clean_name("sai ",last_name="Charan",country="India")       ## mixed arguments (must start at beginning)

clean_name("sai",last_name="charan",)                      ## Default parameter

```

```python
def clean_name(name):
    if not name:
        return None
    else:
        cleaned = name.strip().lower()
        return cleaned
cln_name = clean_name("Maria")    
print(cln_name)
```

```python
def clean_name(name):
    lo_cleaned = name.strip().lower()
    up_cleaned = name.strip().upper()
    return lo_cleaned, up_cleaned
lo_name, up_name = clean_name(" Maria ")
print(lo_name)
print(up_name)
```

### Transformation functions

```python
## Task : clean an email and split it into username and domain
def clean_and_split_email(email):
    cl_email = email.strip().lower()
    username, domain = cl_email.split("@")
    return {"username": username,
            "domain":domain}
print(clean_and_split_email("Computer.support@gmail.com"))
```

### Validation Function

```python
## Task : check if a password meets the minimum length of 8
def is_valid_password(password):
    return len(password) >= 8
print(is_valid_password("1232425456"))
```

```python
## Check whether an email has a basic valid format
def is_valid_email(email):
    return "@" in email and "." in email
is_valid_email("saicharan@gmail.com")
```

### orchestration function

### 1. Recieve an email from the user

### 2. validate the email

### 3. if it is invalid, log an error in a file

### 4. if it is valid, clean and structure the email

### 5. log each steps of program

```python
## Orchestrator Function
def process_user_email(email):
    write_log("app started")
    ## we must check if it is valid
    # is_valid_email(email)

    ## if it is invalid, we must log the problem in a file

    if not is_valid_email(email):
        write_log(f"Invalid Email Recieved: {email}")
   ## if it is valid, we clean it and store structured information
    else:
        clean_email = clean_and_split_email(email)
        write_log(f"Processed Email: {clean_email}")

    write_log("app stopped")

   ## recieve a email from user
email = input("please enter your Email: ")
process_user_email(email)

```
