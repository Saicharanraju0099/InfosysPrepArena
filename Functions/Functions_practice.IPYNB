### Write a function factorial(n) that returns the factorial of a number.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)        ## 5 * (4 * (3 * (2 * (1))))
print(factorial(5))
```

### Write a function count_vowels(text) that counts vowels in a string.

```python
def count_vowels(text):
    count = 0
    vowels = "aeiou"

    for char in text:
        if char in vowels.lower():
            count = count + 1
    return count
print(count_vowels("Saicharan"))
```

### Write a function reverse_string(text) that returns the reversed string.

```python
def reverse_string(text):
    reversed_text = ""
    print(f"Starting with: '{text}'\n") # Initial message

    for char in text:
        # We add the new character to the FRONT
        reversed_text = char + reversed_text
        print(f"Added '{char}' -> Current result: '{reversed_text}'")

    return reversed_text

# Calling the function
final_result = reverse_string("ABC")
print(f"\nFinal reversed string: {final_result}")
```

### Write a function find_max(lst) that returns the largest number in a list.

```python
def find_max(lst):
    if not lst:
        return None
    
    ## Start by assuming the first item is the largest
    largest = lst[0]
    for number in lst:
        if number > largest:
            largest = number
            
    return largest
# 1. Define a list of numbers
my_numbers = [10, 45, 2, 99, 14]

# 2. Call the function and store the result
max_value = find_max(my_numbers)

# 3. Display the result
print(max_value)  # Output: 99

        
```

### Write a function count_even(lst) that counts how many even numbers are in a list.

```python
def count_even(lst):
    count = 0
    for number in lst:
        ## Check if the remainder is 0 when divide by zero
        if number % 2 == 0:
            count = count + 1
    return count
number = [1, 2, 3, 4, 5, 6]
print(count_even(number))
```

### Write a function remove_duplicates(lst) that returns a list with duplicates removed.

```python
def remove_duplicates(lst):
    unique_list = []
    for item in lst:
        if item not in unique_list:
            unique_list.append(item)
    return unique_list
numbers = [2,4,3,5,7,6,2,3,1,8,9,2]
print(remove_duplicates(numbers))
```

### Write a function sum_list(lst) that returns the sum of all numbers in a list.

### Write a function second_largest(lst) that finds the second largest element.

```python
def second_largest(lst):
    largest = lst[0]
    second_largest = lst[0]
    for num in lst:
        if num > largest:
            second_largest = largest
            largest = num
        elif num > second_largest:
            second_largest = num
    return second_largest

numbers = [20,30,40,90]
print(second_largest(numbers))
```

### Write a function is_prime(n) that checks whether a number is prime.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
print(is_prime(17))
    
```

### Write a function palindrome(word) that checks if a string is a palindrome.

```python
def palindrome(word):
    word = word.strip().lower().replace(" ","")
    n = len(word)
    for i in range(n // 2):
        if word[i] != word[n - 1 - i]:
            return False
    return True
word = "Was it a car or a cat I saw"
print(palindrome(word))
```

### Write a function fibonacci(n) that returns the first n Fibonacci numbers.

```python
def fibonacci(n):
    ## start with first two numbers 
    seq = [0,1]
    ## Add numbers until we have 'n' of them
    for i in range(n-2):
        seq.append(seq[-1] + seq[-2])
    return seq[0:n] 
print(fibonacci(5))
```

### Write a function digit_sum(num) that returns the sum of digits of a number.

```python
def digit_sum(num):
    total = 0
    # Use abs() to handle negatives and str() to make it iterable
    for digit in str(abs(num)):
        total += int(digit)
    return total

# Example:
print(digit_sum(502)) 
```

### Write a function calculator(a, b, operator) that performs:

### addition

### subtraction

### multiplication

### division

```python
def calculator(a,b, operator):
    if operator == '+':
        return a + b
    elif operator == '-':
        return a - b
    elif operator == '*':
        return a * b
    elif operator == '/':
        if b == 0:
            return "Error: Division by zero"
        return a / b
    else:
        return "Invalid Operator"

print(calculator(10,2, '+'))
```

### Write a function grade(score) that returns:

### A → 90+

### B → 80–89

### C → 70–79

### D → 60–69

### F → below 60

```python
def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

print(grade(85))
```

### Write a function count_frequency(lst) that counts frequency of each element.

### Example output:

### [1,2,2,3,3,3]
### {1:1, 2:2, 3:3}

```python
def count_frequency(lst):
    counts = {}
    for item in lst:
        if item in counts:
            counts[item] += 1
        else:
            counts[item] = 1
    return counts
print(count_frequency([1,2,3,4,1,2,6,2,4,3,6]))
```

### Write a function matrix_sum(matrix) that returns the sum of all elements in a 2D list.

```python
def matrix_sum(matrix):
    total = 0
    for row in matrix:
        for element in row:
            total = total + element
    return total

my_matrix = [
           [1,2,3],
           [4,5,6]
]

print(matrix_sum(my_matrix))
```
