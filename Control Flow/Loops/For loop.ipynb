### for loop

```python
for i in (1,2,3,4,5):
    print(i)
```

```python
items = (1,2,3,4,5)
for item in items:
    print(f"Round: {item}")
```

```python
# Sequence List
items = [1,2,3,4,5,"code"]
for item in items:
    print(f"Round: {item}")
```

```python
items = "ANALYTICS"
for item in items:
    print(f"Round: {item}")
```

```python
## range()
for item in range(2,10,2):
    print(f"Round: {item}")
```

```python
## for loop use cases
scores = [80,50,60,75]
total = 0
for score in scores:
    total += score
    print("Current Total:", total)
print("Final Total", total)
```

```python
## we use for loop to transform data like cleaning data before processing
files = [' Report.csv','DATA.csv',' final.TXT']
for file in files:
    file = file.strip().lower().replace('txt','csv')
    print(f"Processing {file}")

```

```python
# print the multiplication table
for item in range(1,11):
    result = 7 * item 
    print(f"7 x {item} = {result}")
    
```

```python
for outeritemloop in range(1,7):     # Outer loop : Rows 1 to 6
    for inneritemloop in range(outeritemloop):  # inner loop : print 'outeritemloop' stars means 1 to 6 with increment
        print("*", end=' ')     
    print()                      ## move to next line after each row
    
```
