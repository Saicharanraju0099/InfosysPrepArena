### loop through a list of days and print only the working days, skipping the weekends

```python
days = ['Sunday','Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
weekends = ['Saturday','Sunday']
for day in days:
    if day in weekends:
        continue
    print(f"workdays : {day}")
```

### scan emails to block unsafe data from entering the system

```python
emails = [
        'data@gmail.com',
        'data@outlook.com',
        'data@yahoo.com',
        'DROP TABLE USERS;',
        'data@support.com'
]
for email in emails:
    if "DROP TABLE" in email:
        print("Suspicious email detected!")
        break
    print("Processing email:", email)
```

### check whether any file name appears more than once 
**print("Duplicate found") if a duplicate exists,otherwise print "All files are unique"**

```python
file_list = [
    'report.csv',
    'data.xlsx',
    'summary.docx',
    'report.csv',
    'data.csv'
]
seen_files = set()
for file in file_list:
    if file in seen_files:
        print(f"{file} duplicate found" )
        break
    seen_files.add(file)
else:
    print("all files are unique")
```

### Nested Loops

```python
for x in range(3):   ## outer loop beginning
    for y in range(2):   ## inner loop beginning
        for z in range(2): # nested inner loop beginning
            print(f"x value ({x}), y value({y}), z value({z})")  
```

```python
## crossing data
colors = ['red', 'green', 'blue']
sizes  = ['S', 'M', 'L']
for color in colors:
    for size in sizes:
        print(f"{color} - size {size}")

```

```python
## Navigate hierarchy of nested for loops
years = [2026,2027]
months = ['Jan','Feb','Mar']
days = range(1,32)
for year in years:
    for month in months:
        for day in days:
            print(f"report_{year}_{month}_{day}.csv")
```

```python
## SELECT count(*) FROM customers where id IS NULL;
tables = ['customers', 'orders', 'products']
columns = ['id', 'create_date']
for t in tables:
    for c in columns:
        print(f"SELECT count(*) FROM {t} where {c} IS NULL;")
```
