### Print numbers from 10 to 1 using a for loop.

```python
for i in range(10, 0, -1):
    print(i) 
```

### Print all even numbers between 1 and 50.

```python
for even_number in range(1,50):
    if even_number%2 == 0:
        print(even_number)
```

### Count how many numbers between 1 and 100 are divisible by 3.

```python
count = 0
for i in range(1, 101):
    if i % 3 == 0:
       count = count + 1
print(count)
```

### Thinking in Patterns

**Print this pattern**

### *
### **
### ***
### ****
### *****

```python
for i in range(1, 6):      # Outer loop handles the number of rows (1 to 5) ROWS (Vertical movement)
    for j in range(i):     # Inner loop handles the number of stars in each row COLUMNS (Horizontal movement)
        print("*", end=" ") # Prints star followed by a space, stays on same line
    print()                # Prints a newline to move to the next row
```

### Reverse a string using a for loop only.

```python
text = "Python"
reversed_text = ""

for char in text:
    reversed_text = char + reversed_text  # Adds new char to the BEGINNING

print(reversed_text)  

    
    
```

### Count vowels in a string.

```python
vowels = "AEIOU"
text = "SAICHARAN"
count = 0   ## Initialize count to 0 before the loop
for char in text:
    if char in vowels:
        count = count + 1    # Add 1 to the counter
print(count)
```

### Print a 4x4 square of stars.

```python
for i in range(4):                ## outer loop rows
    for j in range(4):            ## inner loop columns
        print("*", end = " ")     ## print star and stay on same line
    print()                       ## print new line after inner loop ends
```

### Print this pattern
### *
### * *
### * * *
### * * * *

```python
# Outer loop: Iterates from 1 to 4 to create 4 rows
for i in range(1, 5):
    # Inner loop: Prints '*' based on the current row number (i)
    for j in range(i):
        # Print star on the same line with a space
        print("*", end=" ")
    
    # Move to the next line after each row is finished
    print()
```

### Print numbers from 1 to 50 but skip multiples of 5.

### Logic Strengthening

### Find the largest number in a list (without using max()).

```python
list = [1,2,3,7,9,15,4,10]
largest = list[0]  
for num in list:
    if num > largest:
        largest = num
print(largest)
    
```

### Count how many times a specific number appears in a list.

```python
numbers = [1, 3, 5, 3, 7, 3, 9]
target = 3

count = 0

for num in numbers:
    if num == target:
        count += 1

print("Count:", count)
```
