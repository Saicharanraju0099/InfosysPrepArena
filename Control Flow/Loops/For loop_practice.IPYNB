### Check if a number is prime (use only for loop logic).

```python
def prime_numbers(n):
    if n < 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
prime_numbers(11)
```

### Remove duplicates from a list using only loops and conditionals.

```python
numbers = [2,3,4,2,3,5,6,7,8,2,3,4,2]
unique_numbers = []
for num in numbers:
    found = False
    for unique in unique_numbers:
        if num == unique:
            found = True
            break
    if not found:
        unique_numbers.append(num)
print(unique_numbers)
```

Print a right-aligned triangle:
###        *
###       **
###      ***
###     ****
###    *****

```python
for i in range(1, 5 + 1):  # loop to control each row
    for j in range(5 - i):  # loop to print spaces before stars
        print(" ", end="")  # print a space without moving to the next line
    for k in range(i):  # loop to print stars in each row
        print("*", end="")  # print star without going to the next line
    print()  # move to the next line after printing stars
```

### Print a pyramid:
###    *
###   ***
###  *****
### *******
### *********

```python
rows = 5      ## total number of rows in the pyramid
for i in range(1, rows + 1):  # loop to control each row
    for j in range(rows - i):  # loop to print spaces before stars
        print(" ", end = " ")    # print space without moving to the next line
    for k in range(2 * i -1):    # loop to print stars (1,3,5,7,9...)
        print("*", end = " ")     # print star without going to next line  
    print()    # move to the next line after each row
```

### Second Largest Number
### From a list of numbers, find the second largest number without using sort() or max().

```python
numbers = [10,20,30,25,40,50,35,15] 
largest = numbers[0]    # Initialize largest to the first element of the list
second_largest = numbers[0]  # Initialize second largest to the first element of the list
for num in numbers:
    if num > largest:
        second_largest = largest     # Update second largest before updating largest
        largest = num                # Update largest to the new number
    elif num > second_largest and num != largest:
        second_largest = num         # Update second largest if current number is between largest and second largest

print("Largest number is:", largest)
print("Second largest number is:", second_largest)
```

### Character Frequency
### Given a string "programming", count how many times each character appears.

```python
string = 'programming'  # store the given string

for char in string:  # loop through each character in the string
    count = 0  # variable to count how many times the character appears

    for length in string:  # loop again through the string to compare characters
        if char == length:  # check if the characters are the same
            count = count + 1  # increase the count if they match

    print(f"{char}: {count}")  # print the character and its frequency
```

### Reverse Words in Sentence
### Input:
### "python is powerful"
### Output:
### "powerful is python"

```python
sentence = "python is powerful"
words = sentence.split() # Split the sentence into words
reversed_sentence = "" # Initialize an empty string to store the reversed sentence
for i in range(len(words) - 1, -1, -1): # Loop through the list of words in reverse order
    reversed_sentence += words[i] + " "  # Add each word to the reversed sentence with a space
print(reversed_sentence.strip())  # Print the reversed sentence, using strip() to remove any extra space at the end
```

### Sum of Digits
### Given num = 48291, calculate the sum of its digits using a for loop.

```python
number = 48291
sum = 0
for digit in str(number): # Convert the number to a string to iterate through each digit
    sum = sum + int(digit)    # Add the integer value of each digit to the sum
print("Sum of digits in", number, "is:", sum)
```
