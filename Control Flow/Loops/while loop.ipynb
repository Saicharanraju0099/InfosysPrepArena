### while loop

```python
count = 1       # Initialize count to 1
while count <= 5:    # Loop until count is greater than 5
    print(count)
    count += 1       # Increment count by 1 in each iteration
```

```python
answer = ''
while answer.lower() != 'yes':  # Loop until the user types 'yes'
    answer = input("Do you want to continue? (yes/no): ")
    if answer.lower() == 'yes':
        print("Continuing...")
    else:
        print("Please type 'yes' to continue.")
```

```python
while True:  # Infinite loop
    user_input = input("Type 'exit' to stop the loop: ")
    if user_input.lower() == 'exit':
        print("Exiting the loop.")
        break  # Exit the loop when the user types 'exit'
    else:
        print("You typed:", user_input)
```

### allow up to 3 attempts
### if the user types "yes",print"Glad we are on the same page"
### otherwise, print "3 strikes, you. are out!"

```python
attempts = 0
while attempts < 3: # Infinite loop
    answer = input("Do you agree ? (yes/no): ")
    if answer.lower() == 'yes':
        print("Gald we are on the same page")
        break    # Exit the loop when the user types 'yes'
    attempts += 1
else:
        print("3 strikes, you are out!")
```
