### Validate the quality and corrections of email values
**Must not be empty**

**Must contain '.' and '@'**

**Must contain exactly one '@' symbol**

**Must end with '.com', '.org', or '.net'**

**Must not be longer than 254 characters**

**Must start with or end with letter or digit**

```python
email = "computer.support@gmail.com"
valid = True
## Clean the string and check if it's empty
email = email.strip()
if email == "":
    print("Email cannot be empty")
    valid = False
if "." not in email or "@" not in email:
    print("Email must contain '.' and '@'")
    valid = False
if email.count("@") != 1:
    print("Email must contain only one '@'")
    valid = False
if not email.endswith((".com", ".org", ".net")):
    print("Email must end with '.com', '.org', or '.net'")
    valid = False
if len(email) > 254:
    print("Email cannot be longer than 254 characters")
    valid = False
if not (email[0].isalnum() and email[-1].isalnum()):
    print("Email must start and end with an alphanumeric character")
    valid = False
if valid:
    print("Email is Valid")
```

### Challenge 2
### Validate the quality and correctness of passwords
**Must not be empty**

**Must be at least 8 characters**

**Must include at least 1 Uppercase**

**Must include at least 1 Lowercase**

**Must not be same as the email**

**Must contain one special character**

**Must not contain any spaces**

**Must start with and end with letter or digit**

```python
password = "Computer$123" 
email = "computersupport@gmail.com"
valid = True
# Password Must not be Empty
Password = password.strip()
if Password == "":
    print("Password cannot be empty")
    valid = False
# Password Must be at least 8 characters long
if len(password) < 8:
    print("Password must be at least 8 characters long")
    valid = False
# Must contain at least one uppercase letter and lowercase letter
if not any(c.isupper() for c in password) or not any(c.islower() for c in password):
    print("Password must contain at least one uppercase and one lowercase letter")
    valid = False
## Must not same as email
if password.lower() == email.lower():
    print("password cannot be the same as email")
    valid = False
# Must contain at least one special character
if not any(not c.isalnum() for c in password):
    print("Password must contain at least one special character")
    valid = False
## Must not contain any white spaces inside password
if any(c.isspace() for c in password):
    print("Password cannot contain any white spaces")
    valid = False
# Must start with or end with letter or digit
if not (password[0].isalnum() and password[-1].isalnum()):
    print("Password must start and end with a letter or digit")
    valid = False

if valid:
    print("password is valid")
```
