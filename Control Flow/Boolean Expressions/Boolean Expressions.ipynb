### Boolean Expressions
### A Boolean expression in Python is any expression that evaluates to either `True` or `False`.


```python
## Values
print(True)
print(False)
print(type(True))
print(type(False))
```

## Functons

```python
print(bool(1))    ## True if the value is non-zero or non-empty
print(bool(0))    ## False if the value is zero or empty
print(bool(""))   ## False if the string is empty
print(bool(None)) ## False if the value is None
```

```python
email = ""
phone = "123-456-7890"
username = ""
# allow registration if email or phone or username is provided
print(any([email, phone, username]))  ## any() returns True if at least one element is truthy
# allow registration only if email and phone and username are provided
print(all([email, phone, username]))  ## all() returns True if all elements are truthy
```

```python
## isinstance() # checks if a value is of a specific datatype
print(isinstance(5, int))        ## True if value is of type int
print(isinstance("hello", str))  ## True if value is of type str
print(isinstance(5, str))        ## False if value is not of type str
```

```python
## endswith() # checks if a string ends with a specific suffix
print("hello world".endswith("world"))  ## True if string ends with "world"
print("hello world".startswith("h"))  ## True if string starts with "hello"
```

### Comparision Operators
### It compares two values and return True or False based on the result

```python
print(10 == 10) ## True if values are equal
print(10 != 5)  ## True if values are not equal
print(10 > 5)   ## True if left value is greater than right value
print(10 < 20)  ## True if left value is less than right value 
print(10 >= 10) ## True if left value is greater than or equal to right value
print(10 <= 20) ## True if left value is less than or equal to right value
print(1 < 4 < 10) ## True if 1 is less than 4 and 4 is less than 10
```

```python
## Strings Comparison
print("apple" == "apple")  ## True if strings are identical
print("a" < "b")       ## True if "a" comes before "b" in alphabetical order
print("apple" < "banana") ## True if "apple" comes before "banana" in alphabetical order
```

### Logical Operations


```python
print(3 > 1 and 5 < 2)  ## True if both conditions are true
print(3 > 1 or 5 < 2)   ## True if at least one condition is true
print(not (3 > 1))      ## True if condition is false
```

```python
## Check if the system is under pressure based on CPU and Memory usage
cpu_usage = 85  ## CPU usage in percentage
memory_usage = 95  ## Memory usage in percentage
print(cpu_usage > 80 or memory_usage > 90)  ## True if both CPU and Memory usage are high
```

```python
## Checking User Creditials before login
email = True
password = False
print(email and password)  ## True if both email and password are provided
```

### not operator reverse the truth.It turns true into false and false into true

```python
print(not(3 > 2))      # False because 3 is greater than 2, so the condition is true, and not True is False
print(not(True))       # False because not True is False
print(not(False))      # True because not False is True
```

```python
## Chained Comparison Operators
## Allow access if the user is logged in or they are guest but they must not be banned

is_logged_in = True
is_guest = False
is_banned = True
print((is_logged_in or is_guest) and not is_banned)  ## True if user is logged in or they are guest but not banned
```
